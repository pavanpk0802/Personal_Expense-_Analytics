```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 300

df = pd.DataFrame({
    "Expense_ID": range(1, n + 1),
    "Date": pd.date_range("2025-01-01", periods=n),
    "Category": np.random.choice(
        ["Food", "Transport", "Shopping", "Bills", "Entertainment", "Healthcare"],
        n
    ),
    "Payment_Method": np.random.choice(
        ["Cash", "UPI", "Card"], n
    ),
    "Amount": np.random.randint(100, 10000, n)
})

df.to_csv("personal_expenses_raw.csv", index=False)

print("Dataset created successfully!")
print("Records:", len(df))

display(df.head())
```

```python
# Inspect the dataset
print("Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

# Convert date
df["Date"] = pd.to_datetime(df["Date"])

# Remove invalid expenses
df = df[df["Amount"] > 0]

# Reset index
df = df.reset_index(drop=True)

# Save cleaned data
df.to_csv("personal_expenses_cleaned.csv", index=False)

print("\nData cleaning completed!")
print("Final Shape:", df.shape)
```

```python
import matplotlib.pyplot as plt

category_spending = (
    df.groupby("Category")["Amount"]
    .sum()
    .sort_values(ascending=False)
)

print(category_spending)

plt.figure(figsize=(9, 6))
category_spending.plot(kind="bar")

plt.title("Total Spending by Category")
plt.xlabel("Expense Category")
plt.ylabel("Total Amount")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

```python
# Create month column
df["Month"] = df["Date"].dt.to_period("M").astype(str)

# Calculate monthly spending
monthly_spending = df.groupby("Month")["Amount"].sum()

print(monthly_spending)

# Create chart
plt.figure(figsize=(10, 6))

monthly_spending.plot(kind="line", marker="o")

plt.title("Monthly Spending Trend")
plt.xlabel("Month")
plt.ylabel("Total Spending")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
payment_spending = (
    df.groupby("Payment_Method")["Amount"]
    .sum()
    .sort_values(ascending=False)
)

print(payment_spending)

plt.figure(figsize=(8, 5))

payment_spending.plot(kind="bar")

plt.title("Spending by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Total Spending")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
# Calculate KPIs
total_spending = df["Amount"].sum()
average_expense = df["Amount"].mean()
total_transactions = len(df)
highest_expense = df["Amount"].max()
highest_category = category_spending.idxmax()

print("PERSONAL EXPENSE KPIs")
print("=" * 30)
print(f"Total Spending      : ₹{total_spending:,.2f}")
print(f"Average Expense     : ₹{average_expense:,.2f}")
print(f"Total Transactions  : {total_transactions}")
print(f"Highest Expense     : ₹{highest_expense:,.2f}")
print(f"Highest Spend Category: {highest_category}")

# Final dashboard
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Category
category_spending.plot(kind="bar", ax=axes[0, 0])
axes[0, 0].set_title("Spending by Category")
axes[0, 0].set_xlabel("Category")
axes[0, 0].set_ylabel("Amount")

# Monthly trend
monthly_spending.plot(kind="line", marker="o", ax=axes[0, 1])
axes[0, 1].set_title("Monthly Spending Trend")
axes[0, 1].set_xlabel("Month")
axes[0, 1].set_ylabel("Amount")
axes[0, 1].tick_params(axis="x", rotation=45)

# Payment method
payment_spending.plot(kind="bar", ax=axes[1, 0])
axes[1, 0].set_title("Spending by Payment Method")
axes[1, 0].set_xlabel("Payment Method")
axes[1, 0].set_ylabel("Amount")

# Expense distribution
axes[1, 1].hist(df["Amount"], bins=10, edgecolor="black")
axes[1, 1].set_title("Expense Amount Distribution")
axes[1, 1].set_xlabel("Expense Amount")
axes[1, 1].set_ylabel("Number of Transactions")

plt.suptitle("Personal Expense Analytics Dashboard", fontsize=18)
plt.tight_layout()

plt.savefig("personal_expense_dashboard.png", dpi=300, bbox_inches="tight")
plt.show()
```
